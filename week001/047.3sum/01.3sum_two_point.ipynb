Given an integer array nums, return all the triplets `[nums[i], nums[j], nums[k]]` such that `i != j`, `i != k`, and `j != k`, and `nums[i] + nums[j] + nums[k] == 0`.

Notice that the solution set must not contain duplicate triplets.

```python
def threeSum(nums):
    store = [] # chứa danh sách có tổng 3 số bằng 0
    nums.sort() # sắp xếp lại list

    if len(nums) < 3 or nums[0] + nums[1] + nums[2] > 0 or nums[-1] + nums[-2] + nums[-3] < 0:
        return store
    
    
    for idx in range(len(nums)-2):
        if idx > 0 and nums[idx] == nums[idx - 1]:  # Bỏ qua phần tử trùng lặp với phần tử trước đó
            continue
        
        if nums[idx] > 0: # Nếu phần tử cố định đã lớn hơn 0, không thể có tổng bằng 0 => thoát ngay
            break
        
        left =  idx + 1
        right = len(nums) - 1

        while left < right:
            total = nums[idx] + nums[left] + nums[right]

            if total == 0:
                store.append([nums[idx], nums[left], nums[right]])

                # Bỏ qua các phần tử trùng lặp tiếp theo
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1
                    
                # Di chuyển cả hai con trỏ để tìm bộ ba tiếp theo
                left += 1
                right -= 1

            elif total < 0:
                left += 1  # Tăng tổng lên nếu nhỏ hơn 0
            else:
                right -= 1  # Giảm tổng xuống nếu lớn hơn 0

    return store
```

```python
nums = [-1,0,1,2,-1,-4]
threeSum(nums)
```

```python
nums = [0,1,1]
threeSum(nums)
```

```python
nums = [0,0,0]
threeSum(nums)
```
