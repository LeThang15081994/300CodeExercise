Given an `m x n` binary matrix mat, return the distance of the nearest `0` for each cell.

The distance between two adjacent cells is `1`.

```python
def updateMatrix(matrix):
    row = len(matrix)
    col = len(matrix[0])

    dist_matrix  = [[float('inf')] * col for _ in range(row)] # khởi tạo ma trận kết quả.

    queue = []  # khởi tạo hàng đợi
    
    directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]  # Duyệt 4 hướng

    for nr in range(row):
        for nc in range(col):
            if matrix[nr][nc] == 0:
                dist_matrix[nr][nc] = 0  # Đặt ô 0 trong ma trận kết quả
                queue.append((nr,nc)) # thêm vào hàng đợi toạ độ ô giá trị bằng 0

    while queue:
        x, y = queue.pop(0) # lấy phần tử đầu tiên trong hàng dợi và xoá phần tử đó.
        
        for dx, dy in directions: # duyêt các ô lân cận
            nx = x + dx
            ny = y + dy

            if 0 <= nx < row and 0 <= ny < col: # kiểm tra xem (nx,ny) có nằm trong khoảng ranh giới của ma trận không?
                if dist_matrix[nx][ny] > dist_matrix[x][y] + 1:
                    #Cập nhật khoảng cách và thêm vào hàng đợi
                    dist_matrix[nx][ny] = dist_matrix[x][y] + 1
                    queue.append((nx,ny)) 

    return dist_matrix
```
