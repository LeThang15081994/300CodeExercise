Given an `m x n` binary matrix mat, return the distance of the nearest `0` for each cell.

The distance between two adjacent cells is `1`.

```python
def updateMatrix(matrix):
    row = len(matrix)
    col = len(matrix[0])

    result_matrix  = [[float('inf')] * col for _ in range(row)] # khởi tạo ma trận kết quả.

    for i in range(row):
        for j in range(col):
            if matrix[i][j] == 0:
                result_matrix[i][j] = 0

            else:
                if i > 0 :
                    result_matrix[i][j] = min(result_matrix[i][j], result_matrix[i - 1][j] + 1)
                
                if j > 0 :
                     result_matrix[i][j] = min(result_matrix[i][j], result_matrix[i][j - 1] + 1)

    
    

    return result_matrix
```

```python
mat = [[0, 0, 0],
       [0, 1, 0],
       [0, 0, 0]
       ]
updateMatrix(mat)
```

```python
mat = [
    [0, 0, 0],
    [0, 1, 0],
    [1, 1, 1]
]
updateMatrix(mat)
```
