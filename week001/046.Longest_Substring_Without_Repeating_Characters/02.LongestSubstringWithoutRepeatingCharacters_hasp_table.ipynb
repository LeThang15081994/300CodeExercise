Given a string s, find the length of the longest substring without repeating characters.

```python
def lengthOfLongestSubstring(s):
    left = 0
    store = set()  #Hash table (set) để lưu các ký tự trong cửa sổ
    length = 0

    for idx in range(len(s)): # Duyệt từng ký tự bằng con trỏ idx
        while s[idx] in store:
            store.remove(s[left])  # Xóa ký tự tại vị trí `left`
            left += 1 # Dịch con trỏ trái sang phải

        store.add(s[idx]) # Thêm ký tự hiện tại vào set
        length = max(length, idx - left + 1) # Cập nhật độ dài lớn nhất tìm được
        
    return length
```
