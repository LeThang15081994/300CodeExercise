Given a string s, find the length of the longest substring without repeating characters.

```python
def lengthOfLongestSubstring(s):
    count = 0
    store = []
    length = 0

    # xét trường hợp chuỗi nhỏ hơn hoặc bằng 1 ký tư
    if len(s) == 1:
        return 1
    if not s:
        return 0

    # kiểm tra trong list 'store' đã có ký tự value chưa?
    for value in s:
        if value in store: # trường hợp đã có 
            length = max(length,count) # gán giá trị count đếm độ dài chuỗi vào length
            while store and store[0] != value: # thực hiện reset lại list 'store' và 'count'
                store.pop(0) # Xóa phần tử đầu tiên
                count -=1  # Giảm số lượng phần tử trong cửa sổ

            store.pop(0) # Xóa chính ký tự trùng lặp
            store.append(value) # Thêm lại ký tự hiện tại vào cuối danh sách
            
        else:
            store.append(value) # Thêm lại ký tự hiện tại vào cuối danh sách
            count +=1 # Tăng biến đếm 'count'

    length = max(length,count) # gán giá trị count đếm độ dài chuỗi vào length
    return length
```
