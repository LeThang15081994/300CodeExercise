Given a string s, find the length of the longest substring without repeating characters.

```python
def lengthOfLongestSubstring(s):
    count = 0
    length = 0
    store = []
    
    if not s:
        return 0

    for value in s:
        if value not in store: 
            store.append(value)
            count +=1

        else:
            length = max(length, count)

            while store[0] != value and store:
                store.pop(0)
                count -=1

            if store:
                store.pop(0)

            store.append(value)
            count +=1

    length = max(length,count)

    return length
```
