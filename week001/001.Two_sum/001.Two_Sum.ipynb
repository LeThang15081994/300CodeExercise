Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

hash table

```python
def twoSum(nums, target):
        map = {} # Bảng băm để lưu các số đã duyệt và chỉ mục của chúng
        
        for index, num in enumerate(nums):
            quo = target - num # Số cần tìm để đạt được tổng bằng target
            if quo in map:
                return ([map[quo],index]) # Trả về chỉ mục của cặp số
            else:
                map[num] = index  Lưu số hiện tại và chỉ mục của nó vào bảng băm
```

```python
nums = [2,7,11,15]
target = 9
print(twoSum(nums, target))
```

```python
nums = [3,2,4]
target = 6
print(twoSum(nums, target))
```

```python
nums = [3,3]
target = 6
print(twoSum(nums, target))
```

```python
nums = [2,7,11,15,3,3]
target = 5
print(twoSum(nums, target))
```
