Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

An input string is valid if:

Open brackets must be closed by the same type of brackets.

Open brackets must be closed in the correct order.

Every close bracket has a corresponding open bracket of the same type.

Stack

```python
def isValid(s):
    if len(s) % 2 != 0: # nếu chiều dài chuỗi là lẻ thì trả về false luôn
         return False

    stack=[]  # Khởi tạo stack để lưu các dấu mở
    map = {')': '(', '}': '{', ']': '['} # Bảng ánh xạ các dấu đóng và dấu mở tương ứng
    for char in s:
        # Nếu là dấu mở, thêm vào stack
        if char in map.values():
            stack.append(char)
        # Nếu là dấu đóng, kiểm tra khớp với dấu mở trên cùng của stack
        elif stack and stack[-1] == map[char]:
            stack.pop()  # Bỏ dấu mở khỏi stack nếu khớp
        else:
            return False  # Không khớp, trả về False ngay lập tức

    # Chuỗi hợp lệ nếu stack rỗng sau khi duyệt xong
    return not stack
```

```python
s='()'
print(isValid(s))
```

```python
s='()[]{}'
print(isValid(s))
```

```python
s='(]'
print(isValid(s))
```

```python
s='()(})'
print(isValid(s))
```
