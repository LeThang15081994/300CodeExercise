```python
def isValid(s):
    if len(s) % 2 != 0: # nếu chiều dài chuỗi là lẻ thì trả về false luôn
         return False

    stack=[]  # Khởi tạo stack để lưu các dấu mở
    map = {'(':')', '{':'}' , '[':']'} # Bảng ánh xạ các dấu đóng và dấu mở tương ứng
    for i in s:
        if i in '({[':
            stack.append(i)   # Nếu là dấu mở, thêm nó vào stack
        else:
            if not stack or map[stack.pop()] != i:  # Kiểm tra xem dấu đóng có khớp với dấu mở trên cùng của stack không
                return False
    return not stack # Chuỗi chỉ hợp lệ nếu stack rỗng sau khi duyệt hết
```

```python
s='()'
print(isValid(s))
```

```python
s='()[]{}'
print(isValid(s))
```

```python
s='(]'
print(isValid(s))
```

```python
s='()(})'
print(isValid(s))
```
